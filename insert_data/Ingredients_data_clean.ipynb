```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as plt
```

```python
from sqlalchemy import create_engine
DATABASE_URL = "postgresql://postgres:123456@localhost:5432/cosmetic"
engine = create_engine(DATABASE_URL)
```

```python
query = 'SELECT "Ingredients" FROM product_python;'
```

```python
df = pd.read_sql_query(query, engine)
```

```python
def to_replace (symbols, ingredients):
    for symbol in symbols:
        item = ingredients.replace(symbol, ', ')
        ingredients= item
    return item
```

```python
symbols_to_replace = ['&', '**', '/', ', (and)']
df['new']= df['Ingredients'].apply(lambda x:to_replace(symbols_to_replace,x))
```

```python
ingredients_list = [x for x in df['new'].apply(lambda x:x.split(', '))]
ingredients = sum(ingredients_list, [])
```

```python
clear_ingredients =[x.strip('*#%! .') for x in ingredients]
clear_ingredients =[x.strip('*#%! .+') for x in clear_ingredients]
```

```python
unique_ingredients = list(set(clear_ingredients))
ingredients_df= pd.DataFrame(unique_ingredients)
```

```python
import re

def has_pattern(text,pattern=()):
    return bool(re.search(pattern, text))
```

```python
ingredients_df['clean']=ingredients_df[0].apply(lambda x: has_pattern(x, r'\$'))
ingredients_df.loc[3525,0]='potassium sorbate'
```

```python
ingredients_df['clean']=ingredients_df[0].apply(lambda x: has_pattern(x, r'\!'))
ingredients_df=ingredients_df.drop(3961)
```
